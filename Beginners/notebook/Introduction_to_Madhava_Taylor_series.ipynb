## Euler's formula:

### **Expressing Euler's number ($e$) as a $\text{{\bf{Madhava}} {\bf{Taylor}} {\bf{series}}}$:**

$\text{Consider a } \bf{Madhava} \text{ } \bf{Taylor} \text{ } \bf{series} \text{ of the following form: }$

$f({\bf{x}}) = a_0+a_1{\bf{x}}+a_2{\bf{x}}^2+a_3{\bf{x}}^3+...+a_{n}{\bf{x}}^n$

$\text{The derivatives of the function }f({\bf{x}})\text{ can be written as follows: }$

$\frac{d(f({\bf{x}}))}{d{\bf{x}}} = a_1+{{2}{\cdot}a_2{\cdot}{\bf{x}}}+{{3}{\cdot}a_3{\cdot}{\bf{x}}^2}+...+{n{\cdot}a_{n}{\cdot}{\bf{x}}^{n-1}}$

$\frac{d^2(f({\bf{x}}))}{d{\bf{x}}^2} = {{2}{\cdot}a_2}+{{(3.2)}{\cdot}a_3{\cdot}{\bf{x}}}+...+{(n.(n-1)){\cdot}a_{n}{\cdot}{\bf{x}}^{n-2}}$

$\vdots$

$\frac{d^n(f({\bf{x}}))}{d{\bf{x}}^n} = {{(n.(n-1).(n-2)...3.2.1)}\cdot{a_{n}}} = {{n!}\cdot{a_{n}}}$

$\because \frac{d(f({\bf{x}}))}{d{\bf{x}}} = f({\bf{x}}) \text{ for } f({\bf{x}}) = e^{\bf{x}}$

$ \text{ By evaluating the function at } {\bf{x}}=0 \text{, we get: }$

$e^{\bf{0}} = 1, \text{ } \therefore a_0 = 1$

$\frac{d(f({\bf{0}}))}{d{\bf{x}}} = a_1+{{2}\cdot{a_2\cdot{\bf{0}}}}+{{3}\cdot{a_3\cdot{\bf{0}}^2}}+...+{{n}\cdot{a_{n}\cdot{\bf{0}}^{n-1}}} \implies e^{\bf{0}} = a_1 \implies a_1 = 1$

$\frac{d^2(f({\bf{0}}))}{d{\bf{x}}^2} = {{2}\cdot{a_2}}+{{3.2}\cdot{a_3\cdot{\bf{0}}}}+...+{{n.(n-1)}\cdot{a_{n}\cdot{\bf{0}}^{n-2}}} \implies e^{\bf{0}} = {{2}\cdot{a_2}} \implies a_2 = \frac{1}{2}$

$\vdots$

$\frac{d^n(f({\bf{0}}))}{d{\bf{x}}^n} = {{(n.(n-1).(n-2)...3.2.1)}\cdot{a_{n}}} = {{n!}\cdot{a_{n}}} \implies e^{\bf{0}} = {n!}\cdot{a_{n}} \implies a_{n} = \frac{1}{n!}$

$\therefore e^{\bf{x}} \text{ can be rewritten as a power series: }$

$e^{\bf{x}} = 1 + \frac{{\bf{x}}}{1!} + \frac{{\bf{x}}^2}{2!} + \frac{{\bf{x}}^3}{3!} + ... = {\sum}\limits_{n=0}^{n \to \infty}\frac{{\bf{x}}^n}{n!}$

#### Helper classes and functions:

````python
class Factorial():
    def __init__(self, verbose=False):
        self.verbose = verbose
        self.cache_recall = True

    @cache
    def n_factorial(self, n):
        '''https://docs.python.org/3/library/functools.html'''
        return n * self.n_factorial(n-1) if n else 1

class Euler_Madhava_Taylor_Series(Factorial):
    def __init__(self):
        super().__init__()
        self.x = None

    @cache
    def compute_euler_madhava_taylor_series(self, num_steps):
        if self.verbose and num_steps:
            print(f'Number of Madhava Taylor series computational steps remaining: {num_steps - 1} ...', end = '\r', flush=True)
        elif self.verbose:
            print('Finished computing Madhava Taylor series ...' + ' '*25, end = '\r', flush=True)
        self.cache_recall = False
        return self.compute_euler_madhava_taylor_series(num_steps-1) + ((num_steps / self.n_factorial(num_steps) * math.pow(self.x, num_steps))) if num_steps else 0

class Exponent(Euler_Madhava_Taylor_Series):
    def __init__(self, n=5, **kwargs):
        super().__init__()
        self.num_steps = n
        for k in kwargs:
            try:
                _ = getattr(self, str(k))
                setattr(self, str(k), kwargs[k])
            except:
                raise ValueError(f'Attribute {str(k)} not available ...')

    def __call__(self, x):
        self.x = x
        self.cache_recall = True
        assert self.num_steps >= 0 and isinstance(self.num_steps, int), f'Expected value for n to be an integer, instead received: {self.num_steps} ...'
        return self.compute_euler_madhava_taylor_series(self.num_steps)
````

#### Checking the impact on execution time from the use of unbounded function cache:

The unbounded function cache is defined using the decorator ```@cache``` from the built-in module ```functools```.

This a form of computational time optimization technique by returning the values of a function, should the same inputs repeat again over time, during the execution of a program.

There is an increased memory foot-print for executing the program than when executing without the cache, in exchange for a faster code execution.

This technique is also known as: [memoization](https://en.wikipedia.org/wiki/Memoization). 

[Python implemention of this cache using the built-in module: ```functools```](https://docs.python.org/3/library/functools.html), is threadsafe; so that the wrapped function can be multi-threaded, by ensuring a coherent underlying data structure while performing concurrent updates; during muti-threaded code execution.

It should be noted that the assumption of code execution performance gains during the usage of unbounded function cache is made by applying the generally true idea that the computational steps are typically more time consuming than the memory operations.

In real world use cases though, such performance guarantees cannot be assumed; since there can be scenarios where the memory operations are sometimes slower than the computational steps being optimized.

````python
Euler_Number.num_steps = int(5e2)
start_time = time.time()
Euler_Number.verbose = False # True #
print(Euler_Number(x))
exe_t_3 = time.time() - start_time
print(f'Finished execution in: {exe_t_3} ...')
````

````python
exe_t_delta = exe_t_1 - exe_t_3
````

````python
msg_prefix = 'Unbounded function cache based performance optimization'
msg_suffix = 'memoization thy name is irony'
fail_msg_1 = 'Unbounded function cache based performance optimization is not working ...'
fail_msg_2 = 'Unbounded function cache based performance optimization not initialized ...'

print(
    f'{msg_prefix} is working with a {exe_t_delta}s faster execution ...' if exe_t_3 < exe_t_1 and Euler_Number.cache_recall else fail_msg_1
)
print(
    f'{fail_msg_2} ...' if not Euler_Number.cache_recall else ''
)
print(
    f'{msg_prefix} is slower than computational steps by {-exe_t_delta}s, {msg_suffix} ...' if exe_t_3 > exe_t_1 and Euler_Number.cache_recall else ''
)
print(
    f'{msg_prefix} took the same time as the computational steps, {msg_suffix} ...' if exe_t_3 == exe_t_1 and Euler_Number.cache_recall else ''
)
````

### **Basic properties of the function $e^{i{\bf{x}}}$:**

$\frac{d(e^{ix})}{dx} = \frac{d(e^{ix})}{d(ix)} \cdot \frac{ix}{d(x)} = ie^{ix} \because \frac{d(e^x)}{dx} = e^x$

$\therefore \frac{d(e^{ix})}{dx} = -sin(x) + i \cdot {cos(x)} \implies \frac{d(e^{ix})}{dx} = i \cdot (cos(x) + i\cdot{sin(x)})$

$Alternatively, \because e^{ix} = cos(x) + i.sin(x) \implies \frac{d(e^{ix})}{dx} = \frac{d(cos(x) + i.sin(x))}{dx} = -sin(x) + i \cdot {cos(x)}$

$\therefore \frac{d^2(e^{ix})}{dx^2} = -cos(x) - i \cdot {sin(x)} \implies \frac{d(e^{ix})}{dx} = i^2 \cdot (cos(x) + i\cdot{sin(x)})$

$\therefore \frac{d^3(e^{ix})}{dx^3} = sin(x) - i \cdot {cos(x)} \implies \frac{d(e^{ix})}{dx} = i^3 \cdot (cos(x) + i\cdot{sin(x)})$

$\vdots$

$\therefore \frac{d^n(e^{ix})}{dx^n} = i^ne^{ix}$

#### **Expressing trigonometric sine $(sin(θ))$ function as $\text{{\bf{Madhava}} {\bf{Taylor}} {\bf{series}}}$:**

$\text{Consider a } \bf{Madhava} \text{ } \bf{Taylor} \text{ } \bf{series} \text{ of the following form: }$

$f({\bf{θ}}) = a_0+a_1{\bf{θ}}+a_2{\bf{θ}}^2+a_3{\bf{θ}}^3+...+a_{n}{\bf{θ}}^n \text{, where } f({\bf{θ}}) = sin({\bf{θ}})$

$\text{By evaluating, }sin({\bf{θ}}) \text{ at } θ=0, \because sin({\bf{0}}) = 0 \implies a_0 = 0$

$\text{By evaluating, } \frac{d{(sin({\bf{θ}})})}{dθ} \text{ at } θ=0, \because \frac{d{(sin({\bf{θ}})})}{dθ} = cos({\bf{θ}}) \text{ and } cos({\bf{0}}) = 1 \implies a_1 = 1$

$\text{By evaluating, } \frac{d^2{(sin({\bf{θ}})})}{dθ^2} \text{ at } θ=0, \because \frac{d^2{(sin({\bf{θ}})})}{dθ^2} \implies \frac{d{(cos({\bf{θ}})})}{dθ} = -sin({\bf{θ}}) \text{ and } -sin({\bf{0}}) = 0 \implies {(1.2)}\cdot{a_2} = 0 \therefore a_2=0$

$\text{By evaluating, }\frac{d^3{(sin({\bf{θ}})})}{dθ^3} \text{ at } θ=0, \because \frac{d^3{(sin({\bf{θ}})})}{dθ^3} \implies \frac{d{(-sin({\bf{θ}})})}{dθ} = -cos({\bf{θ}}) \text{ and } -cos({\bf{0}}) = 1 \implies {(1.2.3)}\cdot{a_3} = -1 \therefore a_3 = -\frac{1}{3!}$

$\text{By evaluating, }\frac{d^4{(sin({\bf{θ}})})}{dθ^4} \text{ at } θ=0, \because \frac{d^4{(sin({\bf{θ}})})}{dθ^4} \implies \frac{d{(-cos({\bf{θ}})})}{dθ} = sin({\bf{θ}}) \text{ and } sin({\bf{0}}) = 0 \implies {(1.2.3.4)}\cdot{a_4} = 0 \therefore a_4 = 0$

$\text{By evaluating, }\frac{d^5{(sin({\bf{θ}})})}{dθ^5} \text{ at } θ=0, \because \frac{d^5{(sin({\bf{θ}})})}{dθ^5} \implies \frac{d{(sin({\bf{θ}})})}{dθ} = cos({\bf{θ}}) \text{ and } cos({\bf{0}}) = 1 \implies {(1.2.3.4.5)}\cdot{a_5} = 1 \therefore a_5 = \frac{1}{5!}$

$\text{By evaluating, }\frac{d^6{(sin({\bf{θ}})})}{dθ^6} \text{ at } θ=0, \because \frac{d^6{(sin({\bf{θ}})})}{dθ^6} \implies \frac{d{(cos({\bf{θ}})})}{dθ} = -sin({\bf{θ}}) \text{ and } sin({\bf{0}}) = 0 \implies {(1.2.3.4.5.6)}\cdot{a_6} = 0 \therefore a_6 = 0$

$\text{By evaluating, }\frac{d^7{(sin({\bf{θ}})})}{dθ^7} \text{ at } θ=0, \because \frac{d^7{(sin({\bf{θ}})})}{dθ^7} \implies \frac{d{(-sin({\bf{θ}})})}{dθ} = -cos({\bf{θ}}) \text{ and } -cos({\bf{0}}) = 1 \implies {(1.2.3.4.5.6.7)}\cdot{a_7} = -1 \therefore a_7 = -\frac{1}{7!}$

$\vdots$

$\text{By evaluating, }\frac{d^{(2n)}{(sin({\bf{θ}})})}{dθ^{(2n)}} \text{ at } θ=0, \because \frac{d^{(2n)}{(sin({\bf{θ}})})}{dθ^{(2n)}} = (i)^{(2n)}sin({\bf{θ}}) \text{ and } sin({\bf{0}}) = 0 \implies {(2n)!}\cdot{a_{2n}} = 0 \therefore a_{2n} = 0$

$\text{By evaluating, }\frac{d^{(2n+1)}{(sin({\bf{θ}})})}{dθ^{(2n+1)}} \text{ at } θ=0, \because \frac{d^{(2n+1)}{(sin({\bf{θ}})})}{dθ^{(2n+1)}} = (-1)^{n}cos({\bf{θ}}) \text{ and } cos({\bf{0}}) = 1 \implies {(2n+1)!}\cdot{a_{2n+1}} = (-1)^{n} \therefore a_{2n+1} = \frac{(-1)^{n}}{(2n+1)!}$

$\text{Using generalization, if n } \in (2z), \text{ where }z \in ℤ \text{, the corresponding coefficient } a_n = 0 \implies \text{ For } n \ge 0 \text{, where }n \in ℤ \text{, } a_{2n} = 0$

$\text{Using generalization, if n } \in (2z + 1), \text{ where }z \in ℤ \text{ and } n>0 \text{, the corresponding coefficient } a_n = (-1)^{\frac{n-1}{2}}\frac{1}{n!} \implies \text{ For } n \ge 0 \text{, where }n \in ℤ \text{, } a_{2n+1} = {\frac{{(-1)^n}}{(2n+1)!}}$

$\therefore sin({\bf{θ}}) \text{ can be rewritten as a power series: }$

$sin({\bf{θ}}) = \frac{{\bf{θ}}}{1!} - \frac{{\bf{θ}}^3}{3!} + \frac{{\bf{θ}}^5}{5!} - \frac{{\bf{θ}}^7}{7!} + ... = {\sum}\limits_{n=0}^{n \to \infty}{(-1)^n}\frac{{\bf{θ}}^{(2n+1)}}{(2n+1)!}$

````python
class Sine_Series(Factorial):
    def __init__(self):
        super().__init__()
        self.x = None

    def sin_theta(self, n):
        return math.pow(-1, (n-1)) * math.pow(self.x, (2*(n-1))+1) / self.n_factorial((2*(n-1))+1)

    @cache
    def sin_theta_series(self, num_steps):
        if self.verbose and num_steps:
            print(f'Number of Madhava Taylor series computational steps remaining: {num_steps - 1} ...', end = '\r', flush=True)
        elif self.verbose:
            print('Finished computing Madhava Taylor series ...' + ' '*25, end = '\r', flush=True)
        self.cache_recall = False
        return self.sin_theta_series(num_steps-1) +  self.sin_theta(num_steps) if num_steps else 0

class Sin(Sine_Series):
    def __init__(self, n=5, **kwargs):
        super().__init__()
        self.num_steps = n
        for k in kwargs:
            try:
                _ = getattr(self, str(k))
                setattr(self, str(k), kwargs[k])
            except:
                raise ValueError(f'Attribute {str(k)} not available ...')

    def __call__(self, x):
        self.x = x
        self.cache_recall = True
        assert self.num_steps >= 0 and isinstance(self.num_steps, int), f'Expected value for n to be an integer, instead received: {self.num_steps} ...'
        return self.sin_theta_series(self.num_steps)
````

#### Monte Carlo simulations to verify the value of $sin(θ)$ evaluated using $\text{{\bf{Madhava}} {\bf{Taylor}} {\bf{series}}}$:

The verfication of the value of $sin(\theta)$ evaluated using the function that computes the finite number of steps of the $\text{{\bf{Madhava}} {\bf{Taylor}} {\bf{series}}}$ approximating its value, is done by comparing its output with the output of the built-in ```math.sin``` function, through Monte Carlo simulation consisting of a series of randomly chosen $\theta$ $\in$ $[-\pi, \pi]$, using the built-in ```random.SystemRandom()``` class.

````python
num_mc_eval_steps = 100
eval_tolerance = 1e-8

pass_count = 0
for i in range(num_mc_eval_steps):
    Compute_Sin = Sin(verbose=False)
    theta = random.SystemRandom().uniform(-math.pi, math.pi)
    Compute_Sin.num_steps = int(81)
    start_time = time.time()
    sin_theta = Compute_Sin(theta)
    outcome = math.isclose(sin_theta, math.sin(theta), abs_tol=eval_tolerance)
    if outcome:
        pass_count += 1
    exe_t_1 = time.time() - start_time
    if (i + 1) % int(num_mc_eval_steps * 0.1) == 0: 
        print(f'Finished execution of Monte Carlo simulation step {i+1} in: {exe_t_1}...')

acc = pass_count / num_mc_eval_steps
print('')
print(f'Accuracy: {acc}')
````

#### **Expressing hyperbolic sine $(sinh)$ function as $\text{{\bf{Madhava}} {\bf{Taylor}} {\bf{series}}}$:**

$sinh({\bf{x}}) = {-i}\cdot{sin({i}{{\bf{x}}})}$

$\text{Alternatively, } sinh({\bf{x}}) \text{ can be expressed in terms of } e \text{: } sinh({\bf{x}}) = \frac{e^{\bf{x}}-e^{-{\bf{x}}}}{2}$

$\therefore sinh({\bf{x}}) \text{ can be rewritten as a power series: }$

$sinh({\bf{x}}) = \frac{{\bf{x}}}{1!} + \frac{{\bf{x}}^3}{3!} + \frac{{\bf{x}}^5}{5!}... = {\sum}\limits_{n=0}^{n \to \infty}\frac{{\bf{x}}^{(2n+1)}}{(2n+1)!}$

````python
class Hyperbolic_Sine_Series(Factorial):
    def __init__(self):
        super().__init__()
        self.x = None

    def sinh_theta(self, n):
        return math.pow(self.x, (2*(n-1))+1) / self.n_factorial((2*(n-1))+1)

    @cache
    def sinh_theta_series(self, num_steps):
        if self.verbose and num_steps:
            print(f'Number of Madhava Taylor series computational steps remaining: {num_steps - 1} ...', end = '\r', flush=True)
        elif self.verbose:
            print('Finished computing Madhava Taylor series ...' + ' '*25, end = '\r', flush=True)
        self.cache_recall = False
        return self.sinh_theta_series(num_steps-1) +  self.sinh_theta(num_steps) if num_steps else 0

class Sinh(Hyperbolic_Sine_Series):
    def __init__(self, n=5, **kwargs):
        super().__init__()
        self.num_steps = n
        for k in kwargs:
            try:
                _ = getattr(self, str(k))
                setattr(self, str(k), kwargs[k])
            except:
                raise ValueError(f'Attribute {str(k)} not available ...')

    def __call__(self, x):
        self.x = x
        self.cache_recall = True
        assert self.num_steps >= 0 and isinstance(self.num_steps, int), f'Expected value for n to be an integer, instead received: {self.num_steps} ...'
        return self.sinh_theta_series(self.num_steps)
````

#### Monte Carlo simulations to verify the value of $sinh(θ)$ evaluated using $\text{{\bf{Madhava}} {\bf{Taylor}} {\bf{series}}}$:

The verfication of the value of $sinh(\theta)$ evaluated using the function that computes the finite number of steps of the $\text{{\bf{Madhava}} {\bf{Taylor}} {\bf{series}}}$ approximating its value, is done by comparing its output with the output of the built-in ```math.sinh``` function, through Monte Carlo simulation consisting of a series of randomly chosen $\theta$ $\in$ $[-10, 10]$, using the built-in ```random.SystemRandom()``` class.

````python
num_mc_eval_steps = 100
eval_tolerance = 1e-8

pass_count = 0
for i in range(num_mc_eval_steps):
    Compute_Sinh = Sinh(verbose=False)
    theta = random.SystemRandom().uniform(-10, 10)
    Compute_Sinh.num_steps = int(81)
    start_time = time.time()
    sinh_theta = Compute_Sinh(theta)
    outcome = math.isclose(sinh_theta, math.sinh(theta), abs_tol=eval_tolerance)
    if outcome:
        pass_count += 1
    exe_t_1 = time.time() - start_time
    if (i + 1) % int(num_mc_eval_steps * 0.1) == 0: 
        print(f'Finished execution of Monte Carlo simulation step {i+1} in: {exe_t_1}...')

acc = pass_count / num_mc_eval_steps
print('')
print(f'Accuracy: {acc}')
````

$\text{By evaluating, }\frac{d^5{(cos({\bf{θ}})})}{dθ^5} \text{ at } θ=0, \because \frac{d^5{(cos({\bf{θ}})})}{dθ^5} \implies \frac{d{(cos({\bf{θ}})})}{dθ} = -sin({\bf{θ}}) \text{ and } sin({\bf{0}}) = 0 \implies {(1.2.3.4.5)}\cdot{a_5} = 0 \therefore a_5 = 0$

$\text{By evaluating, } \frac{d^6{(cos({\bf{θ}})})}{dθ^6} \text{ at } θ=0, \because \frac{d^6{(cos({\bf{θ}})})}{dθ^6} \implies \frac{d{(-sin({\bf{θ}})})}{dθ} = -cos({\bf{θ}}) \text{ and } cos({\bf{0}}) = 1 \implies {(1.2.3.4.5.6)}\cdot{a_6} = -1 \therefore a_6=-\frac{1}{6!}$

$\text{By evaluating, }\frac{d^{(2n)}{(cos({\bf{θ}})})}{dθ^{(2n)}} \text{ at } θ=0, \because \frac{d^{(2n)}{(cos({\bf{θ}})})}{dθ^{(2n)}} = (i)^{(2n)}cos({\bf{θ}}) \text{ and } cos({\bf{0}}) = 1 \implies {(2n)!}\cdot{a_{2n}} = (-1)^{n} \therefore a_{2n} = \frac{(-1)^{n}}{(2n)!}$

$\text{By evaluating, }\frac{d^{(2n+1)}{(cos({\bf{θ}})})}{dθ^{(2n+1)}} \text{ at } θ=0, \because \frac{d^{(2n+1)}{(cos({\bf{θ}})})}{dθ^{(2n+1)}} = (-1)^{n}sin({\bf{θ}}) \text{ and } sin({\bf{0}}) = 0 \implies {(2n+1)!}\cdot{a_{2n+1}} = 0 \therefore a_{2n+1} = 0$

$\text{Using generalization, if n } \in (2z), \text{ where }z \in ℤ \text{, the corresponding coefficient } a_n = (-1)^{\frac{n}{2}}\frac{1}{n!} \implies \text{ For } n \ge 0 \text{, where }n \in ℤ \text{, } a_{2n} = {\frac{{(-1)^n}}{(2n)!}}$

$\text{Using generalization, if n } \in (2z + 1), \text{ where }z \in ℤ \text{ and } n>0 \text{, the corresponding coefficient } a_n = 0 \implies \text{ For } n \ge 0 \text{, where }n \in ℤ \text{, } a_{2n+1} = 0$

$\therefore cos({\bf{θ}}) \text{ can be rewritten as a power series: }$

$cos({\bf{θ}}) = 1 - \frac{{\bf{θ}}^2}{2!} + \frac{{\bf{θ}}^4}{4!} - \frac{{\bf{θ}}^6}{6!} + ... = {\sum}\limits_{n=0}^{n \to \infty}{(-1)^n}\frac{{\bf{θ}}^{(2n)}}{(2n)!}$

````python
class Cosine_Series(Factorial):
    def __init__(self):
        super().__init__()
        self.x = None

    def cos_theta(self, n):
        return math.pow(-1, (n-1)) * math.pow(self.x, 2*(n-1)) / self.n_factorial(2*(n-1))

    @cache
    def cos_theta_series(self, num_steps):
        if self.verbose and num_steps:
            print(f'Number of Madhava Taylor series computational steps remaining: {num_steps - 1} ...', end = '\r', flush=True)
        elif self.verbose:
            print('Finished computing Madhava Taylor series ...' + ' '*25, end = '\r', flush=True)
        self.cache_recall = False
        return self.cos_theta_series(num_steps-1) +  self.cos_theta(num_steps) if num_steps else 0

class Cos(Cosine_Series):
    def __init__(self, n=5, **kwargs):
        super().__init__()
        self.num_steps = n
        for k in kwargs:
            try:
                _ = getattr(self, str(k))
                setattr(self, str(k), kwargs[k])
            except:
                raise ValueError(f'Attribute {str(k)} not available ...')

    def __call__(self, x):
        self.x = x
        self.cache_recall = True
        assert self.num_steps >= 0 and isinstance(self.num_steps, int), f'Expected value for n to be an integer, instead received: {self.num_steps} ...'
        return self.cos_theta_series(self.num_steps)
````

#### Monte Carlo simulations to verify the value of $cos(θ)$ evaluated using $\text{{\bf{Madhava}} {\bf{Taylor}} {\bf{series}}}$:

The verfication of the value of $cos(\theta)$ evaluated using the function that computes the finite number of steps of the $\text{{\bf{Madhava}} {\bf{Taylor}} {\bf{series}}}$ approximating its value, is done by comparing its output with the output of the built-in ```math.cos``` function, through Monte Carlo simulation consisting of a series of randomly chosen $\theta$ $\in$ $[-\pi, \pi]$, using the built-in ```random.SystemRandom()``` class.

````python
num_mc_eval_steps = 100
eval_tolerance = 1e-8

pass_count = 0
for i in range(num_mc_eval_steps):
    Compute_Cos = Cos(verbose=False)
    theta = random.SystemRandom().uniform(-math.pi, math.pi)
    Compute_Cos.num_steps = int(81)
    start_time = time.time()
    cos_theta = Compute_Cos(theta)
    outcome = math.isclose(cos_theta, math.cos(theta), abs_tol=eval_tolerance)
    if outcome:
        pass_count += 1
    exe_t_1 = time.time() - start_time
    if (i + 1) % int(num_mc_eval_steps * 0.1) == 0: 
        print(f'Finished execution of Monte Carlo simulation step {i+1} in: {exe_t_1}...')

acc = pass_count / num_mc_eval_steps
print('')
print(f'Accuracy: {acc}')
````

#### **Expressing hyperbolic cosine $(cosh)$ function as $\text{{\bf{Madhava}} {\bf{Taylor}} {\bf{series}}}$:**

$cosh(x) = cos(ix)$

$\text{Alternatively, } cosh(x) \text{ can be expressed in terms of } e \text{: } cosh(x) = \frac{e^x+e^{-x}}{2}$

$\therefore cosh({\bf{x}}) \text{ can be rewritten as a power series: }$

$cosh({\bf{x}}) = 1 + \frac{{\bf{x}}^2}{2!} + \frac{{\bf{x}}^4}{4!} + \frac{{\bf{x}}^6}{6!}... = {\sum}\limits_{n=0}^{n \to \infty}\frac{{\bf{x}}^{(2n)}}{(2n)!}$

````python
class Hyperbolic_Cosine_Series(Factorial):
    def __init__(self):
        super().__init__()
        self.x = None

    def cosh_theta(self, n):
        return math.pow(self.x, 2*(n-1)) / self.n_factorial(2*(n-1))

    @cache
    def cosh_theta_series(self, num_steps):
        if self.verbose and num_steps:
            print(f'Number of Madhava Taylor series computational steps remaining: {num_steps - 1} ...', end = '\r', flush=True)
        elif self.verbose:
            print('Finished computing Madhava Taylor series ...' + ' '*25, end = '\r', flush=True)
        self.cache_recall = False
        return self.cosh_theta_series(num_steps-1) +  self.cosh_theta(num_steps) if num_steps else 0

class Cosh(Hyperbolic_Cosine_Series):
    def __init__(self, n=5, **kwargs):
        super().__init__()
        self.num_steps = n
        for k in kwargs:
            try:
                _ = getattr(self, str(k))
                setattr(self, str(k), kwargs[k])
            except:
                raise ValueError(f'Attribute {str(k)} not available ...')

    def __call__(self, x):
        self.x = x
        self.cache_recall = True
        assert self.num_steps >= 0 and isinstance(self.num_steps, int), f'Expected value for n to be an integer, instead received: {self.num_steps} ...'
        return self.cosh_theta_series(self.num_steps)
````

### **Final derivation of Euler's formula:**

#### **Expressing $e^{\bf{x}},\text{ where x } \in ℂ$ as the sum of two distinct $\text{{\bf{Madhava}} {\bf{Taylor}} {\bf{series}}}$:**

$\text{Extending } e^x \text{ to accomodate complex numbers, the function above can be expressed in terms of complex trignometric function: }$

$e^{\bf{x}} \text{, where x } \in ℂ \text{, can be re-written by substituting }x \text{ with } i\theta \text{ as follows:}$

$e^{i\bf{\theta}} = 1 + \frac{i{\bf{\theta}}}{1!} + \frac{(i{\bf{\theta}})^2}{2!} + \frac{(i{\bf{\theta}})^3}{3!} + ...$

$\text{The series can be written as the sum of two components, one real and the other imaginary.}$ 

$\text{ For } \theta \in ℝ \text{, the real part is when } n \in (2z) \text{, where } z \in ℤ \text{ and the imaginary part is when } n \in (2z+1) \text{, where } z \in ℤ.$

$\therefore e^{i\bf{\theta}} = (1 - \frac{{\bf{\theta}}^2}{2!} + \frac{{\bf{\theta}}^4}{4!} - ...) + (\frac{i{\bf{\theta}}}{1!} - \{i\cdot{\frac{{\bf{\theta}}^3}{3!}}\} + \{i\cdot{\frac{{\bf{\theta}}^5}{5!}}\}-...)$

$\therefore e^{i\bf{\theta}} = (1 - \frac{{\bf{\theta}}^2}{2!} + \frac{{\bf{\theta}}^4}{4!} - ...) + i\cdot{(\frac{{\bf{\theta}}}{1!} - {\frac{{\bf{\theta}}^3}{3!}} + {\frac{{\bf{\theta}}^5}{5!}}-...)}$

$\because (1 - \frac{{\bf{\theta}}^2}{2!} + \frac{{\bf{\theta}}^4}{4!} - ...) = cos(\theta) \text{ and } (\frac{{\bf{\theta}}}{1!} - {\frac{{\bf{\theta}}^3}{3!}} + {\frac{{\bf{\theta}}^5}{5!}}-...) = sin(\theta) \implies e^{i\bf{\theta}} = cos(\theta) + i\cdot{sin(\theta)}$

$\text{Alternatively, } e^{\bf{x}} = cos(x) + sin(x), \text{ where x } \in ℂ$

#### **Expressing $e^{\bf{x}},\text{ where x } \in ℝ$ as the sum of two distinct $\text{{\bf{Madhava}} {\bf{Taylor}} {\bf{series}}}$:**

$\text{Extending this Euler's formula to accomodate real numbers: }$

$\theta \in ℝ \text{ can be re-written as: } i\cdot{(-i\theta)}$

$e^{i\cdot{(-i{\bf{\theta}})}} = cos(-i{\bf{\theta}}) + i\cdot{sin(-i{\bf{\theta}})} \implies cos(i{\bf{\theta}}) - i\cdot{sin(i{\bf{\theta}})}$

$\because sin(i{\bf{θ}}) = \frac{i{\bf{θ}}}{1!} - \frac{(i{\bf{θ}})^3}{3!} + \frac{(i{\bf{θ}})^5}{5!} - \frac{(i{\bf{θ}})^7}{7!} + ...$

$\therefore sin(i{\bf{θ}}) = i\cdot{(\frac{{\bf{θ}}}{1!} + \frac{{\bf{θ}}^3}{3!} + \frac{{\bf{θ}}^5}{5!} + \frac{{\bf{θ}}^7}{7!} + ...)} = {i}\cdot{\sum}\limits_{n=0}^{n \to \infty}\frac{{\bf{θ}}^{(2n+1)}}{(2n+1)!}$

$\because {\sum}\limits_{n=0}^{n \to \infty}\frac{{\bf{θ}}^{(2n+1)}}{(2n+1)!} = sinh(\theta) \implies sin(i{\bf{θ}}) = i\cdot{sinh(\theta)}$

$\because cos(i\theta) = (1 - \frac{(i{\bf{\theta}})^2}{2!} + \frac{(i{\bf{\theta}})^4}{4!} - \frac{(i{\bf{\theta}})^6}{6!} +...)$

$\therefore cos(i\theta) = (1 + \frac{{\bf{\theta}}^2}{2!} + \frac{{\bf{\theta}}^4}{4!} + \frac{{\bf{\theta}}^6}{6!} +...) = {\sum}\limits_{n=0}^{n \to \infty}\frac{{\bf{θ}}^{(2n)}}{(2n)!}$

$\because {\sum}\limits_{n=0}^{n \to \infty}\frac{{\bf{θ}}^{(2n)}}{(2n)!} = cosh(\theta) \implies cos(i\theta) = cosh(\theta)$

$\therefore e^{i\cdot{(-i{\bf{\theta}})}} = cosh(\theta) - i\cdot{(i\cdot{sinh(\theta)})}$

$\therefore e^{\bf{\theta}} = cosh(\theta) + sinh(\theta)$

$\text{Alternatively, } e^{\bf{x}} = cosh(x) + sinh(x), \text{ where x } \in ℝ$

````python
Compute_Cosh = Cosh(verbose=False)
Compute_Sinh = Sinh(verbose=False)
Compute_Cosh.num_steps = int(81)
Compute_Sinh.num_steps = int(81)
cosh_theta = Compute_Cosh(1)
sinh_theta = Compute_Sinh(1)

output_dict = dict()
output_dict['Hyperbolic cosine value at θ=1'] = math.cosh(1)
output_dict['Hyperbolic sine value at θ=1'] = math.sinh(1)
output_dict['Estimated hyperbolic cosine value at θ=1'] = cosh_theta
output_dict['Estimated hyperbolic sine value at θ=1'] = sinh_theta
output_dict['Value of the cosh(θ) + sinh(θ) at θ=1'] = math.sinh(1) + math.cosh(1)
output_dict['Estimated Value of the cosh(θ) + sinh(θ) at θ=1'] = cosh_theta + sinh_theta

_ = list(map(lambda k : print(k, ' '*(50-len(k)) + ' : ', output_dict[k]), output_dict))
````

#### Monte Carlo simulations to verify the value of $e^θ$ evaluated using $\text{{\bf{Madhava}} {\bf{Taylor}} {\bf{series}}}$:

The verfication of the value of $e^{\theta}$ evaluated using the function that computes the finite number of steps of the $\text{{\bf{Madhava}} {\bf{Taylor}} {\bf{series}}}$ to approximate the identity: $e^{\theta} = cosh(\theta) + sinh(\theta)$, is done by comparing its output with that of the built-in ```math.exp``` function, through Monte Carlo simulation consisting of a series of randomly chosen $\theta$ $\in$ $[-10, 10]$, using the built-in ```random.SystemRandom()``` class.

````python
num_mc_eval_steps = 100
eval_tolerance = 1e-8

pass_count = 0
for i in range(num_mc_eval_steps):
    Compute_Cosh = Cosh(verbose=False)
    Compute_Sinh = Sinh(verbose=False)
    theta = random.SystemRandom().uniform(-10, 10)
    Compute_Cosh.num_steps = int(81)
    Compute_Sinh.num_steps = int(81)
    start_time = time.time()
    cosh_theta = Compute_Cosh(theta)
    sinh_theta = Compute_Sinh(theta)
    outcome = math.isclose(cosh_theta + sinh_theta, math.exp(theta), abs_tol=eval_tolerance)
    if outcome:
        pass_count += 1
    exe_t_1 = time.time() - start_time
    if (i + 1) % int(num_mc_eval_steps * 0.1) == 0:
        print(f'Finished execution of Monte Carlo simulation step {i+1} in: {exe_t_1}...')

acc = pass_count / num_mc_eval_steps
print('')
print(f'Accuracy: {acc}')
````
